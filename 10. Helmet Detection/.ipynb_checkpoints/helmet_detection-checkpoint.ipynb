```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
# Model training shuru karte hain
results = model.train(
    data='/content/helmet_data/coco128.yaml',
    epochs=25,
    imgsz=640,
    batch=8,
    name='helmet_yolo11n'
)
```

```python
model.save("model_detection_model.pt")
```

```python
model = YOLO('model_detection_model.pt')
```

```python
pred = model.predict("./helmet_img.jpg")
```

```python
pred[0].show()
```

```python
import gradio as gr
from PIL import Image
```

```python
def helmet_detection_model(image):
  pred = model.predict(image)
  predict_image = pred[0].plot()
  return Image.fromarray(predict_image)
```

```python
app = gr.Interface(fn=helmet_detection_model, 
                   inputs="image", 
                   outputs=gr.Image(type="pil"), 
                   title="Helmet Detection")
```

```python
app.launch(show_error=True)
```
