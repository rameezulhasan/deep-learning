```python
import matplotlib.pyplot as plt
```

```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
from IPython import display
display.Image("./image.jpg", width=300, height=300)
```

```python
pred = model.predict("./image.jpg")
```

```python
len(pred)
```

```python
pred[0].show()
```

```python
import matplotlib.pyplot as plt

img = pred[0].plot()

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
from PIL import Image
import gradio as gr
```

```python
def objectDetection(image):
  pred = model.predict(image)
  predict_image = pred[0].plot()
  return Image.fromarray(predict_image)
```

```python
app = gr.Interface(fn=objectDetection, inputs="image", outputs='image', title="Object Detection")
```

```python
app.launch()
```
